```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time
```

```python
service = webdriver.chrome.service.Service('chromedriver.exe')
driver = webdriver.Chrome(service=service)
```

```python
driver.get('https://naver.com')
time.sleep(1)
```

```python
search_box = driver.find_element(By.ID, 'query')
search_box.send_keys('장마')
search_box.send_keys(Keys.RETURN)
time.sleep(1)
```

```python
news_tab = driver.find_element(By.CSS_SELECTOR, "a.tab[href*='where=news']")
news_tab.click()
time.sleep(1)
```

```python
for _ in range(5):
    body = driver.find_element(By.TAG_NAME, 'body')
    body.send_keys(Keys.END)
    time.sleep(1)
```

```python
news_contents = driver.find_elements(By.CSS_SELECTOR, 'span.sds-comps-text-type-headline1')

for news_content in news_contents:
    href_tag = news_content.find_element(By.XPATH, '..')

    title = news_content.text
    href = href_tag.get_attribute('href')

    print(title, " | ", href)
driver.quit()
```
