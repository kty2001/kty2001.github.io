# 차원 축소
- 목적
    - 여러 특성을 2, 3차원으로 줄여 시각화
    - 서로 강하게 관련된 특성 압축
    - 모델 학습에 필요한 특성 수 축소 (계산량 목적)

- PCA: 비지도 차원 축소 (데이터 분산 큰 방향 찾는 축 설계)
- LDA: 지도 차원 축소 (클래스 잘 구분되는 방향 찾는 축 설계)


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris

plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False
```

```python
iris_data = load_iris()

iris_data.keys()
```

```python
iris_df = pd.DataFrame(
    iris_data.data,
    columns=iris_data.feature_names
)

iris_df
```

```python
iris_df['target'] = iris_data.target
iris_df
```

```python
iris_data.target_names
```

```python
iris_df['target_name'] = iris_df['target'].map((
    lambda x: iris_data.target_names[x]
))

iris_df
```

- PCA 적용하기
    - 거리와 분산을 기준으로 계산하기 때문에 PCA 적용 전 스케일링 수행 필요

```python
from sklearn.preprocessing import StandardScaler

X_iris = iris_df.drop(['target', 'target_name'], axis=1)
y_iris = iris_df['target']

scaler = StandardScaler()
iris_scaled = scaler.fit_transform(X_iris)

iris_scaled
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
iris_pca = pca.fit_transform(iris_scaled)

irispca_df = pd.DataFrame(
    iris_pca,
    columns=['pca_component_1', 'pca_component_2']
)

irispca_df['target'] = y_iris
irispca_df['target_name'] = iris_df['target_name']

irispca_df
```

- PCA 설명 분산 비율 확인 (보존율)

```python
pca.explained_variance_ratio_
```

```python
pca.explained_variance_ratio_.sum()
```

96% 정도 보존되어 손실이 4% 정도 발생한 걸로 보여 꽤 잘 압축된 것으로 보임.  
PCA는 정답을 잘 맞추는 방향이 아니라 데이터 분산을 얼마나 많이 보존했는가를 볼 수 있다.

```python
markers = ['^', 'o', 'v']
for i, marker in enumerate(markers):
    x_data = irispca_df[irispca_df['target'] == i]['pca_component_1']
    y_data = irispca_df[irispca_df['target'] == i]['pca_component_2']

    plt.scatter(
        x_data,
        y_data,
        marker=marker,
        label=iris_data.target_names[i]
    )

plt.legend()
plt.xlabel('주성분 1')
plt.ylabel('주성분 2')
plt.title('Iris PCA 결과')
plt.show()
```

2차원 데이터로 축소해서 본다는 것은 모델 성능 올리거나 성능 평가한다기보다는 고차원 데이터를 사람이 볼 수 있는 형태로 줄여 구조 확인하는 시각화 목적

- LDA
    - 클래스가 잘 나뉘는 방향 찾는 목적
    - LDA 최대 차원수: target 클래스 - 1 (iris 경우 2)


```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

lda = LinearDiscriminantAnalysis(n_components=2)
iris_lda = lda.fit_transform(iris_scaled, y_iris)

irislda_df = pd.DataFrame(
    iris_lda,
    columns=['lda_component_1', 'lda_component_2']
)

irislda_df['target'] = y_iris
irislda_df['target_name'] = iris_df['target_name']

irislda_df
```

```python
markers = ['^', 'o', 'v']
for i, marker in enumerate(markers):
    x_data = irislda_df[irislda_df['target'] == i]['lda_component_1']
    y_data = irislda_df[irislda_df['target'] == i]['lda_component_2']

    plt.scatter(
        x_data,
        y_data,
        marker=marker,
        label=iris_data.target_names[i]
    )

plt.legend()
plt.xlabel('LDA 1')
plt.ylabel('LDA 2')
plt.title('Iris LDA 결과')
plt.show()
```

지도학습 분류 문제에서 PCA보다 LDA가 클래스 구분이 더 명확하게 차원축소된다.

| 구분 | PCA | LDA |
| --- | --- | --- |
| 학습 방법 | 비지도 학습 | 지도 학습 |
| 정답 라벨 | 사용 안 함 | 사용함 |
| 목적 | 데이터 분산 최대한 보존 | 클래스 잘 나뉘는 방향 탐색 |
| 주요 사용 | 시각화, 압축, 노이즈 감소 | 분류 전처리, 클래스 분리 시각화 |

- 신용카드 데이터에서 상관관계 높은 변수 줄인다

```python
df = pd.read_excel('data/pca_credit_card.xls', header=1, sheet_name='Data')
df
```

```python
df.info()
```

```python
df = df.rename(
    columns={
        'PAY_0': 'PAY_1',
        'default payment next month': 'default'
    }
)
df = df.drop('ID', axis=1)
df.head()
```

```python
X_data = df.drop('default', axis=1)
y_data = df['default']

print(X_data.shape, y_data.shape)
```

```python
target_summary = pd.DataFrame({
    'count': y_data.value_counts(),
    'ratio': y_data.value_counts(normalize=True)
})
target_summary
```

상관관계 확인
- 상관관계가 높은 여러 변수 확인, 차원 축소 진행

```python
corr = X_data.corr()

plt.figure(figsize=(14, 12))
sns.heatmap(corr, cmap='coolwarm', center=0)
plt.title('상관관계 히트맵')
plt.show()
```

BILL_AMT1 ~ BILL_AMT6는 강한 양의 상관관계 보이기 때문에 차원축소 해도 결과에 큰 차이 없을 것으로 예상

```python
bill_features = ['BILL_AMT1', 'BILL_AMT2', 'BILL_AMT3', 'BILL_AMT4', 'BILL_AMT5', 'BILL_AMT6']
X_bill = X_data[bill_features]
X_bill
```

```python
X_scaler = StandardScaler()
X_bill_scaled = scaler.fit_transform(X_bill)
X_bill_scaled
```

```python
pca_bill = PCA(n_components=2)
X_bill_pca = pca_bill.fit_transform(X_bill_scaled)
X_bill_pca
```

```python
pca_result = pd.DataFrame({
    'component': ['PC1', 'PC2'],
    'explained_variance_ratio': pca_bill.explained_variance_ratio_
})
pca_result
```

```python
pca_bill.explained_variance_ratio_.sum()
```

```python
pca_full = PCA()
pca_full.fit(X_bill_scaled)

cum_ratio = np.cumsum(pca_full.explained_variance_ratio_)

plt.plot(range(1, len(cum_ratio)+1), cum_ratio, marker='o')
plt.xticks(range(1, len(cum_ratio)+1))
plt.xlabel('차원 수')
plt.ylabel('설명력 누적합')
plt.title('설명력 누적합계 그래프')
plt.grid(True)
plt.show()
```

- 차원축소 전후 모델 성능 비교
    - 잘못된 방식: Scaler와 PCA 먼저 학습 진행한 후 trian_test_split 진행한 경우
    -> 테스트 데이터 정보가 전처리 결과에 미리 반영됨
    - 순서
        - train/test 분리
        - train 데이터로 Scaler, PCA 학습
        - test 데이터는 train 데이터로 학습된 전처리 객체로 변환만 진행
        - 모델 학습 및 평가

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

X_train, X_test, y_train, y_test = train_test_split(
    X_bill,
    y_data,
    test_size=0.2,
    random_state=42,
    stratify=y_data
)
print(f'X 학습 데이터: {X_train.shape}')
print(f'X 테스트 데이터: {X_test.shape}')
```

```python
lr_model = LogisticRegression(max_iter=1000)

scaler = StandardScaler()
X_train_before = scaler.fit_transform(X_train)

lr_model.fit(X_train_before, y_train)
X_test_scaled = scaler.transform(X_test)
lr_pred = lr_model.predict(X_test_scaled)

print(f'차원 축소 전 모델의 정확도: {accuracy_score(y_test, lr_pred)}')
print(f'차원 축소 전 모델의 F1: {f1_score(y_test, lr_pred)}')
```

```python
lr_model = LogisticRegression(max_iter=1000)
scaler = StandardScaler()
pca = PCA(n_components=2)

X_train_new = scaler.fit_transform(X_train)
new_pca = pca.fit_transform(X_train_new)

lr_model.fit(new_pca, y_train)
X_test_scaled = scaler.transform(X_test)
X_test_pca = pca.transform(X_test_scaled)

pca_pred = lr_model.predict(X_test_pca)

print(f'차원 축소 후 모델의 정확도: {accuracy_score(y_test, pca_pred)}')
print(f'차원 축소 후 모델의 F1: {f1_score(y_test, pca_pred)}')
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, lr_pred))
print(classification_report(y_test, pca_pred))
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

ConfusionMatrixDisplay.from_predictions(y_test, lr_pred)
plt.title('PCA 전 혼동행렬')
plt.show()

ConfusionMatrixDisplay.from_predictions(y_test, pca_pred)
plt.title('PCA gn 혼동행렬')
plt.show()
```

- PCA는 라벨 없이 데이터의 분산이 큰 방향 찾아 차원 줄인다.
- LDA는 라벨 사용해서 클래스 잘 분리되는 방향으로 차원 줄인다.
- PCA 전 스케일링 필수
- 상관관계 높은 컬럼들은 PCA를 통해 압축한다.
- 성능 비교 시 train/test 데이터 분리 후, train 데이터로만 전처리 객체 학습
- 차원 축소 자체는 성능 향상이 아닌 시각화 및 특성 압축을 통한 계산량 감소에 있다
