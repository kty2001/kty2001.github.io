```python
models = [
    init_chat_model('openai:gpt-5.6-luna'),
    init_chat_model('openai:gpt-4.1-mini'),
    init_chat_model('huggingface:Qwen/Qwen2.5-0.5B-Instruct')
]

user_prompt = '공부를 잘하는 방법'

for model in models:
    model_name = model.model_name if hasattr(model, 'model_name') else model.model_id
    print(f'===={model_name}====')
    ai_message = model.invoke(user_prompt)
    print(ai_message.content)
    print()
```

```python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate.from_template('{product}를 홍보하기 위한 참신한 문구 생성')
print(prompt)

print(prompt.format(product='전기차'))
print(prompt.format(product='카메라'))

llm = init_chat_model('openai:gpt-5.6-luna')
ai_message = llm.invoke(prompt.format(product='카메라'))
ai_message.content
```

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt = ChatPromptTemplate.from_messages([
    ('system', '당신은 {domain} 분야 전문가'),
    ('user', '{question}에 한글로 대답')
])

print(chat_prompt.format(domain='IT', question='langsmith 사용밥'))

llm = init_chat_model('openai:gpt-5.6-luna')
ai_message = llm.invoke(chat_prompt.format(domain='IT', question='langsmith 사용밥'))
ai_message.content
```

```python
from langchain_core.output_parsers import CommaSeparatedListOutputParser

parser = CommaSeparatedListOutputParser()
model_output = '달걀, 고구마, 대파, 쪽파'
parser.parse(model_output)
```

```python
format_instruction = parser.get_format_instructions()
print(format_instruction)
```

```python
prompt = PromptTemplate(
    template = '{subject} 상위 팀 {n}개를 작성 \n(지시사항: {format_instruction})',
    input_variables=['subject', 'n'],
    partial_variables={'format_instruction': format_instruction}
)

print(prompt)
print(prompt.format(subject='kbo', n=7))
```

```python
input = prompt.invoke({'subject': 'kbo', 'n': 7})
print(input)

llm = init_chat_model('openai:gpt-5.6-luna')
ai_message = llm.invoke(input)
print(ai_message.content)

parser = CommaSeparatedListOutputParser()
output = parser.invoke(ai_message)
print(output)
```

```python
chain = prompt | llm | parser
chain.invoke({'subject': 'kbo', 'n': 7})
```

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
chain = llm | parser
chain.invoke('아 너무 덥다. 여름에 먹을 제철과일 제시')
```

```python
from langchain_core.output_parsers import JsonOutputParser

parser = JsonOutputParser()
format_instruction = parser.get_format_instructions()
print(format_instruction)

prompt = PromptTemplate(
    template = '{subject} 분야에서 유명한 책 {n}권 정보 제공\n(지시사항: {format_instruction})',
    input_variables=['subject', 'n'],
    partial_variables={'format_instruction': format_instruction}
)

chain = prompt | llm | parser
result = chain.invoke({'subject': 'computer science', 'n': 4})
```

```python
from pydantic import BaseModel
from langchain_core.output_parsers import PydanticOutputParser
from typing import List

class Book(BaseModel):
    title: str
    author: str
    publisher: str
    year: int = None
    description: str

class BookList(BaseModel):
    books: List[Book]

parser = PydanticOutputParser(pydantic_object=BookList)
format_instruction = parser.get_format_instructions()
print(format_instruction)

prompt = PromptTemplate(
    template = '{subject} 분야에서 유명한 책 {n}권 정보 제공\n(지시사항: {format_instruction})',
    input_variables=['subject', 'n'],
    partial_variables={'format_instruction': format_instruction}
)

chain = prompt | llm | parser
result = chain.invoke({'subject': 'economy', 'n': 4})
result
```

```python
from pprint import pprint
pprint(*result)
```

```python
prompt = PromptTemplate(
    template = '{subject} 분야에서 유명한 책 {n}권 정보 제공)',
    input_variables=['subject', 'n'],
)

chain = prompt | llm.with_structured_output(BookList)
result = chain.invoke({'subject': 'AI', 'n': 4})
result
```

```python
pprint(*result)
```

```python
for book in result.books:
    print(f'제목: {book.title}')
    print(f'저자: {book.author}')
    print(f'출판사: {book.publisher}')
    print(f'연도: {book.year}')
    print(f'설명: {book.description}')
    print()
```
