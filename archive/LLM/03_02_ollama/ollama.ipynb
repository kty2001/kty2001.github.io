````python
import ollama

response = ollama.chat(
    model='llama3',
    messages=[{'role': 'user', 'content': 'Hello World!'}]
)

print(response)
````

````python
def ollama_chat(input):
    response = ollama.chat(
        model='llama3',
        messages=[{'role': 'user', 'content': input}]
    )

    return response.message.content
````

````python
ollama_chat('파이썬으로 퀵소트를 구현해줘.')
````

````python
ollama_chat('''텍스트 요약: **현재 공식 자료 기준으로 보면 Ollama 자체는 개인정보 측면에서 꽤 신뢰할 만한 편**이라고 볼 수 있음. 다만 **"Ollama니까 무조건 안전"은 아님. 핵심은 로컬 모델과 클라우드 모델을 구분하는 것**임.

### 로컬 모델을 쓰는 경우

예를 들어:

```text
ollama run llama3.1
```

처럼 모델을 **PC에 다운로드해서 로컬에서 추론**하면, Ollama 공식 정책상 프롬프트·응답·로컬 모델 상호작용 데이터를 Ollama 서버로 수집하거나 전송하지 않음. ([Ollama][1])

즉:

```text
내 문서
   ↓
Ollama
   ↓
내 PC의 GPU/CPU
   ↓
로컬 LLM
```

구조라서 **입력한 개인정보가 Ollama 서버로 넘어가는 구조가 아님.**

### 다만 주의할 부분

최근 Ollama에는 **Cloud 모델**도 있음.

```text
ollama run llama3.1       → 로컬 모델
ollama run xxx-cloud      → Ollama Cloud
```

Cloud 모델을 사용하면 프롬프트가 서버로 전송되어 추론됨. Ollama는 이 데이터가 저장·로그되거나 모델 학습에 사용되지 않는다고 명시하고 있지만, **어쨌든 외부 서버로 전송되는 것은 로컬 실행과 다름.** ([Ollama][1])

### 개인정보를 최대한 확실하게 차단하려면

Ollama의 **Cloud 기능 자체를 끌 수 있음.**

Windows에서도:

```powershell
$env:OLLAMA_NO_CLOUD=1
```

또는 설정 파일에서 `disable_ollama_cloud: true`를 설정할 수 있음. ([Ollama][2])

그러면 **Ollama를 로컬 전용으로 사용할 수 있음.**

---

### 중국 관련해서는

여기서 **"중국 회사가 아니니까 안전"이라고 판단하면 안 됨.**

중요한 건 국적보다:

**① 소프트웨어가 어디로 데이터를 보내는가
② 어떤 모델을 사용하는가
③ 클라우드 추론인지 로컬 추론인지
④ 오픈소스라면 코드를 검증할 수 있는가**

임.

Ollama의 핵심 코드 자체는 공개 GitHub 저장소에서 확인할 수 있고 MIT 라이선스로 공개되어 있음. ([GitHub][3])

그리고 **Qwen·DeepSeek 같은 중국계 모델을 Ollama에서 로컬로 돌리는 것**도 가능함. 이 경우에도 모델이 중국산이라는 사실과 **네 프롬프트가 중국 서버로 전송되는지 여부는 별개의 문제**임.

**결론:**
개인정보 때문에 Ollama를 사용하는 거라면 **`로컬 모델 + Cloud 기능 비활성화` 조합이면 상당히 보수적인 선택**임. 오히려 이 경우 Ollama보다 **다운로드한 모델 자체와 Ollama에 연결한 외부 프로그램/확장 프로그램**을 같이 확인하는 게 중요함.
''')
````
