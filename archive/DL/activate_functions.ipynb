# 활성화 함수
Sigmoid: 이진 분류 출력층 사용
Tanh: RNN 등 시계열 모델에서 중심화된 출력값 필요 시 사용
ReLU: 대부분 DNN 모델 은닉층에서 기본값 사용
Leaky ReLU: Dead ReLU 문제 보완한 함수

```python
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
```

```python
# Avg = 0, std = 1 100개 데이터에 10 곱해 std 10으로 바꾸고 value만 가져오기
x = torch.sort(torch.randn(100) * 10)[0]
x
```

```python
act = nn.Sigmoid()
print(act(x))
```

```python
torch.sigmoid(x)
```

```python
plt.plot(x.numpy(), torch.sigmoid(x).numpy())
plt.show()
```

# Tanh

```python
act = nn.Tanh()
print(act(x))
```

```python
torch.tanh(x)
```

```python
plt.plot(x.numpy(), torch.tanh(x).numpy())
plt.show()
```

# ReLU

```python
act = nn.ReLU()
act(x)
```

```python
torch.relu(x)
```

```python
plt.plot(x.numpy(), torch.relu(x).numpy())
plt.show()
```

```python
act = nn.LeakyReLU()
act(x)
```

```python
import torch.nn.functional as F

F.leaky_relu(x)
```

```python
plt.plot(x.numpy(), F.leaky_relu(x).numpy())
plt.show()
```
