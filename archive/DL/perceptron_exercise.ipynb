```python
import numpy as np

def perceptron(x1, x2, weights, bias):
    x = np.array([x1, x2])
    w = np.array(weights)
    return int(np.dot(x, w) + bias > 0)
    # return 1 if (w1*x1 + w2*x2 + b) > 0 else 0

# AND Gate
print("AND:", perceptron(1, 1, [0.5, 0.5], -0.7))

# NAND Gate
print("NAND:", perceptron(1, 1, [-0.5, -0.5], 0.7))

# OR Gate
print("OR:", perceptron(1, 0, [0.5, 0.5], -0.2))
```

```python
def weighted_sum(inputs, weights, bias):
    return np.dot(inputs, weights) + bias

inputs = [1, 0]
weights = [0.8, -0.5]
bias = 0.2
output = weighted_sum(inputs, weights, bias)
print("Weighted Sum:", output)
```

```python
from sklearn.neural_network import MLPClassifier

# XOR 문제 해결
X = [[0, 0], [0, 1], [1, 0], [1, 1]]
y = [0, 1, 1, 0]

mlp = MLPClassifier(hidden_layer_sizes=(3, 4, 5, 8, 8), activation='relu', max_iter=3000)
mlp.fit(X, y)

print("Prediction:", mlp.predict(X))
```

activation: 은닉층에서 사용할 활성화 함수
- 입력에 비선형성 추가해 복잡한 패턴 인식하는 함수

- ReLU 계열
    - ReLU: 가장 일반적 사용, 학습 빠름
    - Leaky ReLU: 음수 영역 작은 기울기 허용 (죽은 ReLU 문제 완화)
    - ELU, GELU: Transformer 계열에서 사용
- tanh: 출력 범위가 -1 ~ 1로 나오는게 의미있을 때 사용
    - 입/출력이 0 중심 대칭
    - RNN 계열 (LSTM, GRU)
- logistic: sigmoid (초기 신경망)

- 출력층에서 사용할 활성화 함수
이진 분류: sigmoid (0 ~ 1) -> 확률로 해석
다중 분류: softmax (0 ~ 1, 총합 1) -> 클래스별 확률로 해석
회귀: linear (없음) -> 예측값 그대로 사용
