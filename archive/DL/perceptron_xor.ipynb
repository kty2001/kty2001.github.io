# 단층 퍼셉트론 XOR

```python
import torch
import torch.nn as nn
import torch.optim as optim

import numpy as np

x = torch.tensor([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=torch.float32)
y = torch.tensor([[0], [1], [1], [0]], dtype=torch.float32)

print(x.dtype, y.dtype)

model = nn.Sequential(
    nn.Linear(2, 1),
    nn.Sigmoid()
)

model
```

```python
criterion = nn.BCELoss()
```

```python
optimizer = optim.SGD(
    model.parameters(),
    lr=0.001
)
```

```python
for epoch in range(5000):
    optimizer.zero_grad()

    output = model(x)
    loss = criterion(output, y)

    loss.backward()
    optimizer.step()

    if epoch % 500 == 0:
        print(loss)
        print(f'Epoch {epoch} / Loss: {loss.item():.4f}')
```

```python
with torch.no_grad():
    probability = model(x)
    prediction = (probability >= 0.5).float()

    print(f'probability: {probability}')
    print(f'prediction: {prediction}')
```

단층 퍼셉트론 모델로는 XOR 문제 해결 불가

# 다층 퍼셉트론 XOR

```python
import torch
import torch.nn as nn
import torch.optim as optim

import numpy as np

x = torch.tensor([[0, 0], [0, 1], [1, 0], [1, 1]], dtype=torch.float32)
y = torch.tensor([[0], [1], [1], [0]], dtype=torch.float32)

print(x.dtype, y.dtype)

model = nn.Sequential(
    nn.Linear(2, 4),
    nn.Tanh(),
    
    nn.Linear(4, 1),
    nn.Sigmoid()
)

model
```

```python
criterion = nn.BCELoss()
```

```python
optimizer = optim.SGD(
    model.parameters(),
    lr=0.1
)
```

```python
for epoch in range(10000):
    optimizer.zero_grad()

    output = model(x)
    loss = criterion(output, y)

    loss.backward()
    optimizer.step()

    if epoch % 500 == 0:
        print(f'Epoch {epoch} / Loss: {loss.item():.4f}')
```

```python
with torch.no_grad():
    probability = model(x)
    prediction = (probability >= 0.5).float()

    print(f'probability: {probability}')
    print(f'prediction: {prediction}')
```

tanh 활성화함수 relu 계열 함수의 차이는 초기 학습 안정성 높고, gradient 흐름에 도움됨  
깊은 모델로 갈수록 relu 계열이 더 좋음

이진 분류 출력층 Sigmoid (0 ~ 1 사이 확률값) + BCE Loss 손실함수 계산
