```python
import torch

x = torch.FloatTensor(
    [[1, 2],
    [3, 4]]
).requires_grad_(True)
```

requires_grad_(True): 이 텐서에 대해 미분값 구하겠다 (기울기 추적용)

```python
x1 = x + 2
x2 = x - 2
x3 = x1 * x2
y = x3.sum()

print(x1)
print(x2)
print(x3)
print(y)
```

```python
y.backward()
```

```python
print(x.grad)
```

y = sum((x + 2)(x - 2)) = sum(x^2 -4)

delta(-4)/delta(x) = 0

$
\frac{\partial}{\partial x_i}(x_i^2) = 2x_i
$  
따라서 x.grad는 2x

```python
print(x)    # 텐서값 변경 없음. 역전파는 기울기 계산만 하고 수정 안 함
```
