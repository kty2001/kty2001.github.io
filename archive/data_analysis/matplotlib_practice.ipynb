```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)

# 데이터 준비
x = np.arange(1, 11)
y = np.random.randint(10, 100, 10)

data = np.random.normal(50, 15, 100)
```

```python
plt.bar(x, y, color='b')
plt.plot(x, y, color='r')
plt.hist(data, bins=30, color='g')
plt.scatter(x, y, color='black')
```

```python
months = ['January', 'February', 'March', 'April', 'May', 'June', 
          'July', 'August', 'September', 'October', 'November', 'December']
avg_temperatures = [-1, 0, 5, 10, 15, 20, 25, 23, 19, 12, 5, 0]
rainfall = [50, 40, 60, 80, 100, 120, 150, 140, 110, 90, 70, 60]
math_scores = [85, 90, 88, 76, 95, 89, 72, 84, 91, 87]
english_scores = [78, 82, 85, 80, 88, 90, 84, 79, 85, 83]
```

```python
plt.plot(months, avg_temperatures)
```

```python
plt.bar(months, rainfall)
```

```python
plt.boxplot([math_scores, english_scores], tick_labels=['math_scores', 'english_scores'])
```

```python
plt.hist(math_scores, bins=3)
```

```python
plt.figure()
plt.plot(months, avg_temperatures)
plt.bar(months, rainfall)
plt.boxplot([math_scores, english_scores], tick_labels=['math_scores', 'english_scores'])
plt.hist(np.array(math_scores)/10, bins=3)
plt.show()
```
