```python
import numpy as np

# 5개 지점의 월별 사과, 바나나, 오렌지 판매량 (5x3x6 배열)
# 5개의 지점, 3개의 과일(사과, 바나나, 오렌지), 6개월치 데이터
sales_data = np.array([
    # 지점 1
    [[120, 150, 200, 230, 180, 210],  # 사과 판매량
     [100, 140, 170, 190, 160, 180],  # 바나나 판매량
     [90, 120, 160, 140, 130, 150]],  # 오렌지 판매량

    # 지점 2
    [[80, 90, 110, 140, 130, 120],
     [60, 70, 90, 100, 110, 95],
     [50, 80, 100, 110, 90, 85]],

    # 지점 3
    [[130, 140, 160, 180, 170, 190],
     [110, 120, 130, 150, 140, 160],
     [100, 110, 120, 130, 120, 140]],

    # 지점 4
    [[100, 130, 150, 170, 160, 180],
     [90, 100, 110, 120, 130, 125],
     [80, 110, 140, 130, 120, 110]],

    # 지점 5
    [[90, 110, 130, 140, 120, 130],
     [70, 85, 100, 110, 105, 95],
     [60, 90, 110, 100, 90, 95]]
])

sales_info = ['사과', '바나나', '오렌지']
```

```python
print(np.mean(sales_data, axis=2))
```

```python
print((np.sum(sales_data, axis=(0))))
print(np.argmax(np.sum(sales_data, axis=0), axis=0))
print(list(sales_info[i] for i in np.argmax(np.sum(sales_data, axis=0), axis=0)))
```

```python
print(sales_info[np.argmax(np.sum(sales_data, axis=(2, 0)))])
```

```python
print(sales_data[sales_data > 150])
print(np.where(sales_data > 150))

for bran_i, fru_i, mon_i in zip(*np.where(sales_data > 150)):
    print(f'{mon_i+1}월 {bran_i}지점, 과일: {sales_info[fru_i]}, 판매: {sales_data[bran_i, fru_i, mon_i]}')
```
