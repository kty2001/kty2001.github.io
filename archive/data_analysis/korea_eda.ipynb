```python
data = df.rename(columns={
    'h20_g3': 'sex',
    'h20_g4': 'birth',
    'h20_eco9': 'job_code',
    'p2002_8aq1': 'income'
})

data = data[['sex', 'birth', 'job_code', 'income']]
data
```

```python
# np.where(조건, 값1, 값2): 조건 참이면 값1, 거짓이면 값2
data['sex'] = np.where(data['sex'] == 1.0, 'M', 'F')
data.head()
```

```python
sex_income['M'] / sex_income['F']
```

```python
# t-test로 통계적 유의성 검증
# equal_var: 등분산 가정
import scipy.stats as stats

sex_m = data[data['sex'] == 'M'].dropna()
sex_f = data[data['sex'] == 'F'].dropna()

stats.ttest_ind(sex_m['income'], sex_f['income'], equal_var=True)
```

t-test 결과 해석  
statistic (t-값): 값이 크면 더 두 집단 평균 차이 큼
p-value: 0에 근접 -> p_value 기준 0.05보다 작다 -> 귀무가설 기각  
남성과 여성 평균 소득에는 유의미한 차이 존재

```python
# age 컬럼 = 2026 - bitrh
data = data.assign(age=2026 - data['birth'])
data
```

```python
data['age'] = data['age'].astype('int')
data
```

```python
# 소득 결측치 제거
data_income = data.dropna(subset=['income'])

# 나이별 평균 소득 계산
data_income = data_income.groupby('age')['income'].mean()

data_income.plot()
```

```python
print(data_income.max())
print(data_income.idxmax())
```

```python
data_income = data_income.sort_values()
data_income
```

```python
data = data.assign(agetype=np.where(
    data['age'] < 30, 'youth',
    np.where(data['age'] < 60, 'middle age', 'senior')
))
data
```

```python
sns.countplot(x='agetype', data=data)
```

```python
data.groupby('agetype')['income'].mean()
```

```python
data.groupby(['sex', 'agetype'])['income'].mean()
```

```python
age_sex_income = data.dropna(subset=['income']).groupby(['sex', 'age'], as_index=False)['income'].mean()
age_sex_income
```

```python
sns.lineplot(x='age', y='income', data=age_sex_income, hue='sex')
```

```python
# unstack(): 인덱스 있던 값 컬럼으로 펼치기
age_mean = data.dropna(subset=['income']).groupby(['age', 'sex'])['income'].mean().unstack()

age_mean['ratio_M_over_F'] = age_mean['M'] / age_mean['F']
age_mean
```

```python
age_mean['ratio_M_over_F'].plot(figsize=(8, 3))
plt.axhline(1, linestyle='--', color='r')
plt.title('연령대별 남녀 소득 비율')
plt.ylabel('M / F')
plt.show()
```

```python
sheet_name = 1332

job_name = pd.read_excel('data\(2025년 20차 한국복지패널조사) 조사설계서-가구용(beta1).xlsx', sheet_name='직종코드(2019 7차 신분류)')

job_name
```

```python
# 소분류 -> job_code, Unnamed: 3 -> job
job_name.rename(columns={'소분류': 'job_code', 'Unnamed: 3': 'job'}, inplace=True)

job_name = job_name[['job_code', 'job']]
job_name
```

```python
# data와 job_name을 job_code 기준으로 매핑
job_df = data.merge(job_name, how='left', on='job_code')
job_df
```

```python
job_income_mean = job_df.dropna(subset=['job', 'income'])
job_income_mean = job_income_mean.groupby('job')['income'].mean()
job_income_mean.to_frame()
```

```python
top10 = job_income_mean.sort_values(ascending=False).head(10).sort_values()

plt.figure(figsize=(8, 5))
top10.plot(kind='barh')
plt.title('상위 10개 직업 평균 소득 랭킹')
plt.xlabel('평균 소득')
plt.show()
```

```python
bot10 = job_income_mean.sort_values(ascending=False).tail(10).sort_values()

plt.figure(figsize=(8, 5))
bot10.plot(kind='barh')
plt.title('상위 10개 직업 평균 소득 랭킹')
plt.xlabel('평균 소득')
plt.show()
```
