```python
import matplotlib.pyplot as plt

# 간단한 선 그래프 그리기
plt.plot([1, 2, 3, 4], [10, 20, 25, 30])
plt.title('Simple Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
fig, axs = plt.subplots(2, 2)

# 각 서브플롯에 그래프 그리기
# 2x2 크기의 서브플롯을 생성하여 각 서브플롯에 서로 다른 그래프 생성
axs[0, 0].plot([1, 2, 3], [1, 2, 3])
axs[0, 1].plot([1, 2, 3], [1, 4, 9])
axs[1, 0].plot([1, 2, 3], [1, 8, 27])
axs[1, 1].plot([1, 2, 3], [1, 16, 81])

plt.show()
```

```python
fig, ax = plt.subplots()

# Axes 객체에서 그래프 그리기
ax.plot([1, 2, 3, 4], [10, 20, 25, 30])
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_title('OOP Graph')

plt.show()
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temperatures = [22, 24, 19, 23, 25, 20, 18]

# 선 그래프 그리기
plt.plot(days, temperatures, marker='o', linestyle='-', color='b')
plt.title('Temperature Variation Over a Week')
plt.xlabel('Day')
plt.ylabel('Temperature (°C)')
plt.grid(True)
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [10, 15, 20, 25, 30]
y2 = [5, 10, 15, 20, 25]

# 그래프 그리기
# plot() 함수를 여러 번 호출하면 여러 개의 선을 하나의 그래프에 그릴 수 있음
plt.plot(x, y1, label='Dataset 1', marker='o')
plt.plot(x, y2, label='Dataset 2', marker='s')

# 그래프 꾸미기
plt.title('Customized Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)
plt.show()
```

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [150, 200, 180, 220, 170]

# 막대 그래프 그리기
plt.bar(months, sales, color='skyblue')
plt.plot(months, sales, color='r', marker='^')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales Amount')
for i, m in enumerate(months):
    plt.text(m, sales[i] + 5, sales[i])
plt.show()
```

```python
import matplotlib.pyplot as plt

# 데이터 준비
classA = [1, 85, 88, 90, 93, 95, 97, 99]
classB = [78, 80, 82, 85, 87, 89, 91]
classC = [70, 72, 75, 77, 79, 81, 83]

# 박스플롯 그리기
data = [classA, classB, classC]
plt.boxplot(data, tick_labels=['Class A', 'Class B', 'Class C'])
plt.title('Exam Score Distribution by Class')
plt.ylabel('Score')
plt.show()
```

```python
import numpy as np
heights = np.random.normal(170, 5, 200)  # 평균 170cm, 표준편차 5cm인 데이터 200개 생성

# 히스토그램 그리기
plt.hist(heights, bins=10, color='green', edgecolor='r')
plt.title('Height Distribution of Students')
plt.xlabel('Height (cm)')
plt.ylabel('Number of Students')
plt.show()
```

```python
np.random.seed(0)
study_hours = np.random.normal(5, 1.5, 100)  # 평균 5시간, 표준편차 1.5시간
scores = study_hours * 10 + np.random.normal(0, 5, 100)  # 공부 시간에 따른 점수

# 산점도 그리기
plt.scatter(study_hours, scores, color='purple', alpha=0.6)
plt.title('Relationship between Study Hours and Exam Scores')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 12, 15, 18, 20]

# 그래프 그리기
plt.plot(x, y, color='red', linestyle='--', marker='v')
plt.title('Customized Line Style')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.show()
```

```python
x = [1, 2, 3, 4]
y1 = [10, 20, 25, 30]
y2 = [30, 25, 20, 10]
y3 = [20, 30, 10, 40]

# 다양한 선 스타일로 그래프 그리기
plt.plot(x, y1, linestyle='-', label='Solid Line')
plt.plot(x, y2, linestyle='--', label='Dashed Line')
plt.plot(x, y3, linestyle=':', label='Dotted Line')

# 레이블 및 타이틀 추가
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Line Styles Example')

# 범례 추가
plt.legend()

# 그래프 보여주기
plt.show()
```

```python
x = [1, 2, 3, 4, 5]

# Y축 데이터 설정
y1 = [1, 4, 9, 16, 25]
y2 = [25, 16, 9, 4, 1]
y3 = [20, 3, 10, 4, 5]

# 첫 번째 선: 빨간 실선, 투명도 설정
plt.plot(x, y1, label='y = x^2', color=(0.1, 0.2, 0.5), alpha=0.5)

# 두 번째 선: 파란 점선, 투명도 설정
plt.plot(x, y2, label='y = -x^2', color='#FF5733', alpha=0.7)
plt.plot(x, y3, color=(100/255, 100/255, 100/255), alpha=0.1)

# 레이블 및 타이틀 추가
plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Multiple Lines with Transparency')

# 범례 추가
plt.legend()

# 그래프 보여주기
plt.show()
```

```python
x = np.linspace(0, 10, 100)

# 서브플롯 생성
plt.figure(figsize=(10, 5))

# 첫 번째 그래프
plt.subplot(1, 2, 1)
plt.plot(x, np.sin(x))
plt.title('Sine Wave')

# 두 번째 그래프
plt.subplot(1, 2, 2)
plt.plot(x, np.cos(x))
plt.title('Cosine Wave')

plt.tight_layout()
plt.show()
```

```python
data = np.random.rand(10, 10)

# imshow() 함수로 cmap='viridis' 컬러맵 적용
plt.imshow(data, cmap='viridis')

# 컬러바 추가 (색상과 값의 대응 관계를 나타내는 컬러바)
plt.colorbar()

# 타이틀 추가
plt.title('2D Data with Viridis Colormap')

# 그래프 보여주기
plt.show()
```

```python
data = np.random.rand(10, 10)

# 두 개의 서브플롯에서 다른 컬러맵 적용
plt.subplot(1, 2, 1)
plt.imshow(data, cmap='plasma')
plt.title('Plasma Colormap')
plt.colorbar()

plt.subplot(1, 2, 2)
plt.imshow(data, cmap='inferno')
plt.title('Inferno Colormap')
plt.colorbar()

# 서브플롯 간격 조정
plt.tight_layout()

# 그래프 보여주기
plt.show()
```

```python
from matplotlib.colors import LinearSegmentedColormap

colors = ['#0000FF', '#00FF00', '#FF0000']  # 파란색에서 녹색, 빨간색으로 변환
custom_cmap = LinearSegmentedColormap.from_list('custom', colors)

# 2D 배열 데이터 생성
data = np.random.rand(10, 10)

# 사용자 정의 컬러맵 적용
plt.imshow(data, cmap=custom_cmap)
plt.colorbar()

# 타이틀 추가
plt.title('Custom Colormap')

# 그래프 보여주기
plt.show()
```
