```python
import pandas as pd

data = [
    ['Name', 'Age', 'City', 'Salary'],
    ['Alice', 25, 'New York', 70000],
    ['Bob', 30, 'Los Angeles', 80000],
    ['Charlie', 35, 'Chicago', 120000],
    ['David', 40, 'Houston', 90000],
    ['Eve', 28, 'San Francisco', 75000]
]

df = pd.DataFrame(data[1:], columns=data[0])
```

```python
print(df)
```

```python
print(df['Name'])
print(df['City'])
print(df[df['Age'] >= 30])
```

```python
df['Country'] = 'USA'
df.loc[5] = ['Frank', 33, 'Seattle', 85000, 'USA']
print(df)
```

```python
print(df.drop('Salary', axis=1, inplace=True))
print(df.drop(3, axis=0, inplace=True))
print(df)
```

```python
new_salary = [70000, 80000, 120000, 75000, 85000]
df['Salary'] = new_salary
print(df)
```

```python
df.loc[df['Age'] >= 35, 'Salary'] = df[df['Age'] >= 30]['Salary']*1.05
print(df)
```

```python
sales_data = {
    'OrderID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'Product': ['Laptop', 'Mouse', 'Laptop', 'Monitor', 'Mouse', 'Keyboard', 'Laptop', 'Mouse', 'Monitor', 'Keyboard'],
    'Quantity': [1, 2, 1, 1, 2, 1, 1, 3, 1, 1],
    'Price': [1200, 25, 1200, 300, 25, 100, 1200, 25, 300, 100],
    'CustomerID': [1001, 1002, 1001, 1003, 1004, 1002, 1001, 1003, 1005, 1002],
    'OrderDate': ['2026-01-05', '2026-01-06', '2026-01-07', '2026-01-10', '2024-01-12', '2026-01-15', '2026-01-18', '2026-01-20', '2026-01-22', '2026-01-25']
}
df_sales = pd.DataFrame(sales_data)
df_sales
```

```python
df_sales['Total'] = [q*p for q, p in zip(df_sales['Quantity'], df_sales['Price'])]
id_group = df_sales.groupby('CustomerID')['Price'].sum()
id_group.to_frame()
```

```python
labtop_group = df_sales[df_sales['Product'] == 'Laptop'].groupby('Product')['Price'].mean()
labtop_group.to_frame()
```

```python
product_group = df_sales.groupby('Product')['Quantity'].sum()
print(product_group)
product_group.nlargest(1).to_frame()
```

```python
df_sales['OrderDate'] = pd.to_datetime(df_sales['OrderDate'])
date_filter = df_sales[df_sales['OrderDate'] > '2026-01-10']
date_filter
```

```python
pro_group = df_sales.groupby('Product').agg({'Quantity': 'sum', 'Total': 'sum'}).sort_values('Total', ascending=False)
pro_group
```

```python
df_eco['OrderDate'] = pd.to_datetime(df_eco['OrderDate'])
df_eco.dtypes
```

```python
df_eco['TotalRevenue'] = [q*p for q, p in zip(df_eco['Quantity'], df_eco['Price'])]
df_eco
```

```python
import matplotlib.pyplot as plt
df_time = df_eco.set_index('OrderDate')
df_monthly = df_time['TotalRevenue'].resample('ME').sum()
plt.figure(figsize=(10, 5))
df_monthly.plot(kind='line', marker='o', color='royalblue', linewidth=2)

plt.title('월별 총 매출액 추이', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('주문 월 (Order Date)', fontsize=11)
plt.ylabel('총 매출액', fontsize=11)
plt.grid(True, linestyle='--', alpha=0.6)

plt.show()
```
