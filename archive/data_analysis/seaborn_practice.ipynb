```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 데이터셋 로드
titanic = sns.load_dataset('titanic')
titanic.info()
```

```python
titanic.head()
```

```python
sns.barplot(x=['Died', 'Survived'], y=[sum(titanic.survived == False), sum(titanic.survived == True)])
plt.title("ㅁㄴㅇㄹTitanic survived")
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()

sns.countplot(x=titanic.survived)
plt.title("Titanic survived")
plt.xlabel('Survived')
plt.ylabel('Count')
plt.show()
```

```python
sex_sur = titanic.groupby('sex')['survived'].mean().to_frame()

sns.barplot(x=sex_sur.index, y=sex_sur['survived'])
plt.title("Titanic survived per sex Ratio")
plt.xlabel('Sex')
plt.ylabel('Survived Ratio')
plt.show()
```

```python
sns.barplot(x=titanic['class'], y=titanic['survived'], hue=titanic.sex, errorbar=None)
plt.title("Titanic class-survived relation")
plt.xlabel('Class')
plt.ylabel('Survived Ratio')
plt.show()
```

```python
sns.histplot(titanic['age'].dropna(), kde=True, bins=30)
plt.title("Titanic age distribution")
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

```python
# 문제 5
sns.scatterplot(x=titanic.age, y=titanic.fare, hue=titanic.survived)
plt.title("Titanic age-fare relation")
plt.xlabel('Age')
plt.ylabel('fare')
plt.show()
```
