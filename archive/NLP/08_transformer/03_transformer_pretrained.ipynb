```python
from transformers import pipeline

classifier = pipeline('text-classification', model='daekeun-ml/koelectra-small-v3-nsmc')
```

```python
def analyze_sentiment(texts):
    results = classifier(texts)

    label_map = {'0': '부정', '1': '긍정'}

    for text, result in zip(texts, results):
        print(f'입력 문장: {text}')
        print(f'모델 결과: {result}')
        print(f'감정: {label_map[result["label"]]}')
        print(f'확률: {round(result["score"], 4)}')
        print()
```

```python
texts = [
    '재밌고 감동 실화',
    '시간 아깝네',
    '대작같은 졸작',
    'tq'
]

analyze_sentiment(texts)
```

```python
from transformers import pipeline

generator = pipeline('text-generation', model='Qwen/Qwen2.5-0.5B-Instruct')
```

```python
def generate_text(text):
    messages = [
        {
            'role': 'user',
            'content': f'다음 문장을 자연스럽게 이어서 작성해줘. {text}'
        }
    ]

    result = generator(
        messages,
        max_new_tokens = 300,
        do_sample = True,
        temperature = 0.7
    )

    return result[0]['generated_text'][-1]['content']
```

```python
text = '오늘 학원에서 수업을 듣다가'

print(generate_text(text)[0])
```

```python
texts = [
    '인공지능이 발전하게 되면 인간을',
    '인간이 살아가는 이유는',
    '삶이란',
    '한국 사람이 똑똑한 이유는',
    '라이엇 게임즈의 미래는'
]

for text in texts:
    print(f'입력: {text}')
    print(f'출력: {generate_text(text)}')
    print()
```
