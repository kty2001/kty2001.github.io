```python
from dotenv import load_dotenv

load_dotenv()
HF_TOKEN = os.getenv('HF_TOKEN')
```

```python
from transformers import pipeline
```

```python
translator = pipeline('translation', model='Helsinki-NLP/opus-mt-ko-en')
translator
```

```python
translator('와 이제 마지막 시간이다')
```

```python
texts = [
    '와 이제 마지막 시간이다',
    '이제 수업 열심히 들어보자',
    '집이나 가자',
    '낼 벌써 수요일 미쳤다'
]
translator(texts)
```

```python
sentiment_clf = pipeline('sentiment-analysis')
```

```python
sentiment_clf(texts)
```

```python
ko_sentiment_clf = pipeline('sentiment-analysis', model='sangrimlee/bert-base-multilingual-cased-nsmc')
ko_sentiment_clf(texts)
```

```python
zero_shot_clf = pipeline('zero-shot-classification')
```

```python
zero_shot_clf(
    'This is a course about the transformers library.',
    candidate_labels = ['education', 'politics', 'business']
)
```

```python
zero_shot_clf(
    "It has a poison and it's very dangerous",
    candidate_labels = ['capybara', 'tiger', 'snake']
)
```

```python
ko_zero_shot_clf = pipeline('zero-shot-classification', model='joeddav/xlm-roberta-large-xnli')
```

```python
ko_zero_shot_clf(
    '2026년 당신은 얼마나 잘 살고 있나요',
    candidate_labels=['사회', '정치', '경제', '건강']
)
```

```python
text_generator = pipeline('text-generation', model='distilgpt2')
```

```python
text_generator(
    'In the course, we will teach you how to',
    max_length=30,
    num_return_sequences=2
)
```

```python
ko_text_generator = pipeline('text-generation', model='skt/ko-gpt-trinity-1.2B-v0.5')
```

```python
ko_text_generator(
    '오늘 저녁에는',
    max_length=30,
)
```

```python
unmasker = pipeline('fill-mask')
```

```python
unmasker(
    'shut the <mask> up',
    top_k=3
)
```

```python
ner = pipeline('ner', model='dslim/bert-base-NER', grouped_entities=True)
```

```python
ner('My name is Capybara and I work at Playdata in Seoul')
```

```python
qna = pipeline('question-answering')
```

```python
qna(
    question='Where do I work?',
    context='My name is Capybara and I work at company',
)
```

```python
ko_qna = pipeline('question-answering', model='klue/roberta-base')
```

```python
ko_qna(
    question="1931년 어떤 진단을 받았나요?",  # 문맥에서 찾고자 하는 질문
    context="""그러던 1931년, 그는 갑작스럽게 폐결핵을 진단받았다. 이상은 하루에 담배를 50개피 이상 피는 것을 자신의 일과라고 표현했을 정도로 엄청난 골초였다"""  # 답변을 추출할 문맥
)
```
